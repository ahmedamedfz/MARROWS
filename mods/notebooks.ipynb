```python
import pydicom
import numpy as np
from PIL import Image, ImageOps
from pydicom.pixel_data_handlers.util import apply_voi_lut

# Path to your DICOM file
dcm_file = '/Users/ahmadfariz/Downloads/prostat/MRI prostat/PIRADS 3/4529674_20250716/Eko Kridono61.dcm'
ds = pydicom.dcmread(dcm_file)

# Apply VOI LUT if present, otherwise use raw pixel array
if 'VOILUTSequence' in ds or 'WindowWidth' in ds:
    img_array = apply_voi_lut(ds.pixel_array, ds)
else:
    img_array = ds.pixel_array

# Apply rescale if needed (usually for CT/MRI)
intercept = ds.get('RescaleIntercept', 0.0)
slope = ds.get('RescaleSlope', 1.0)
img_array = img_array * slope + intercept

# Normalize image to 0–255 (after VOI LUT and rescaling)
img_normalized = (img_array - np.min(img_array)) / (np.max(img_array) - np.min(img_array)) * 255.0
img_uint8 = img_normalized.astype(np.uint8)

# Handle Photometric Interpretation (MONOCHROME1 means black=high intensity)
if ds.PhotometricInterpretation == "MONOCHROME1":
    img_uint8 = np.invert(img_uint8)

# Convert to PIL image
img = Image.fromarray(img_uint8)

# Pad to 2048x2048 without stretching (keep aspect ratio)
img_padded = ImageOps.pad(img, (2048, 2048), method=Image.Resampling.LANCZOS, color=0, centering=(0.5, 0.5))

# Save
output_path = "output_image_labstyle_2k.png"
img_padded.save(output_path)
print(f"Image saved as {output_path}")
```
